#Customer Predictor

A simple neural network: 3 input features → 8 hidden nodes → 1 output.

```python
import torch
import torch.nn as nn
```

```python
class CustomerPredictor(nn.Module):
    def __init__(self):
        super().__init__()
        self.hidden = nn.Linear(3, 8)
        self.output = nn.Linear(8, 1)

    def forward(self, x):
        x = torch.relu(self.hidden(x))
        return self.output(x)
```

#Tensor basics

```python
simple_tensor = torch.tensor([0.5, 1.0, 2.0])
print(simple_tensor)
print(simple_tensor.shape)
```

#Forward pass

```python
model = CustomerPredictor()
# Create a dummy batch of 4 customers, each with 3 features (e.g., age, income, score)
dummy_input = torch.randn(4, 3)
# Pass the input through the model
predictions = model(dummy_input)

print("Predictions tensor:", predictions)
print("Output shape:", predictions.shape)  # Will be torch.Size([4, 1])
```
